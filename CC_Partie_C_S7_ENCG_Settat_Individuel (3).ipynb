## 1. Régression linéaire

Présentation
Catégorie : apprentissage supervisé, régression. Entrées : p = 10 variables numériques de Diabetes . Sortie : un nombre, la progression de la maladie un an après.

Hypothèse clé : la cible est une combinaison linéaire des variables, avec des erreurs de moyenne nulle et de variance à peu près constante.

Étapes : (1) je mets les données dans une matrice X de taille n × p ; (2) je cherche les coefficients qui rendent la somme des carrés des erreurs la plus petite possible (moindres carrés) ; (3) je prédis avec ŷ = β₀ + β₁x₁ + … + β_p x_p.

Appris : β₀ et les β_j . Hyperparamètre : fit_intercept .

J(β) = (1/n) · Σᵢ ( yᵢ − β₀ − Σⱼ βⱼ xᵢⱼ )²

n : nombre de lignes ; p : nombre de variables ; xᵢⱼ : variable j du patient i ; yᵢ : vraie valeur ; β₀ : constante ; β_j : coefficient de la variable j ; J : coût moyen quadratique



### Contextes et choix
- **Contexte 1 (marketing/finance) :** prévoir le chiffre d'affaires mensuel d'une boutique à partir du budget publicitaire, du prix moyen et du nombre de visiteurs.
- **Contexte 2 (logistique) :** estimer le délai de livraison en jours à partir de la distance, du poids et du nombre d'escales.
- **Structure attendue :** tableau *n × p* de variables numériques et une cible continue.
- **Concurrent :** Ridge (si variables corrélées) ou forêt aléatoire (si non-linéaire).
- **Mal adapté :** relation courbe ou très hétérogène, ou beaucoup de valeurs extrêmes.
- **Métrique :** MAE, parce qu'elle s'exprime dans l'unité de la cible et se lit facilement ; RMSE si les grosses erreurs coûtent cher ; R² pour me comparer à la simple moyenne.

````python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
````

## 2. Ridge

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, régression.
* **Entrées :** $p = 10$ variables numériques de la base *Diabetes*.
* **Sortie :** $1$ variable continue (progression de la maladie).
* **Hypothèse clé :** La relation est principalement linéaire, mais certaines variables explicatives sont colinéaires (fortement corrélées), ce qui rend les estimations des moindres carrés instables.
* **Étapes de calcul :** (1) Standardisation des variables d'entrée. (2) Ajout d'une pénalité $L_2$ (norme quadratique) à la fonction objectif. (3) Résolution analytique ou par descente de gradient de l'équation normale régularisée.
* **Paramètre appris :** Les coefficients du modèle $\beta_j$ et l'ordonnée à l'origine $\beta_0$.
* **Hyperparamètre :** Le coefficient de régularisation $\alpha$ (plus il est grand, plus les poids sont contraints vers zéro).

**Équation de la fonction de coût :**
$$J(\beta) = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - \beta_0 - \sum_{j=1}^{p} \beta_j x_{ij} \right)^2 + \alpha \sum_{j=1}^{p} \beta_j^2$$
* **$n$ :** Nombre d'observations.
* **$p$ :** Nombre de caractéristiques (variables).
* **$y_i$ :** Valeur cible réelle pour le patient $i$.
* **$x_{ij}$ :** Valeur de la caractéristique $j$ pour le patient $i$.
* **$\beta_0, \beta_j$ :** Coefficients appris.
* **$\alpha$ :** Hyperparamètre de pénalité ($L_2$).

### C2 — Schéma original
```
    [ Coefficients non régularisés (Moindres Carrés Ordinaires) ]
                                |
                                | (Pénalisation quadratique L2)
                                v
    [ Alpha augmente ---> Coefficients se rapprochent de 0 (sans jamais s'annuler totalement) ]
                                |
                                v
              [ Modèle régularisé à variance réduite ]
```

### C3 — Contextes et choix
* **Contexte 1 (Finance) :** Évaluation de la valeur d'un portefeuille d'actifs financiers en utilisant de nombreux indices macroéconomiques corrélés entre eux.
* **Contexte 2 (Marketing) :** Modélisation de la performance des ventes à partir des dépenses publicitaires sur différents canaux (TV, Web, Réseaux Sociaux), qui augmentent souvent ensemble (forte colinéarité).
* **Structure de données :** Tableau $n \times p$ de caractéristiques quantitatives colinéaires et une variable cible continue.
* **Algorithme concurrent :** Lasso (si on cherche à éliminer complètement les variables inutiles) ou ElasticNet.
* **Situation d'inadaptation :** Si le vrai modèle ne comporte que 2 ou 3 variables réellement importantes parmi des centaines de variables de bruit (Lasso serait bien plus adapté pour forcer la parcimonie).
* **Métrique de performance :** MAE ou RMSE (en évaluant l'erreur moyenne de prédiction de la progression du diabète).

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Comment évoluent les coefficients quand alpha augmente ?*
Quand l'hyperparamètre $\alpha$ augmente, la pénalité sur la magnitude des coefficients $\beta$ devient plus stricte. En conséquence, les coefficients sont progressivement "réduits" (shrinkage) et tendent tous vers $0$, sans jamais atteindre exactement zéro. Cela permet de diminuer la variance globale du modèle en acceptant une légère augmentation du biais.

````python
from sklearn.linear_model import Ridge
import numpy as np

# Comparaison des coefficients pour différentes valeurs d'alpha
for a in [0.1, 1.0, 10.0, 100.0]:
    modele_ridge = make_pipeline(StandardScaler(), Ridge(alpha=a)).fit(Xr_ent, yr_ent)
    coefs = modele_ridge[-1].coef_
    mae = mean_absolute_error(yr_test, modele_ridge.predict(Xr_test))
    print(f"Alpha = {a:5} | MAE = {mae:.2f} | Norme L2 des coefs = {np.linalg.norm(coefs):.3f} | Max Coef = {np.max(np.abs(coefs)):.3f}")
````

## 3. Lasso

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, régression.
* **Entrées :** $p = 10$ variables de la base *Diabetes*.
* **Sortie :** $1$ variable continue.
* **Hypothèse clé :** Seul un petit sous-ensemble de variables d'entrée est réellement pertinent pour expliquer la cible (hypothèse de parcimonie/sparcité).
* **Étapes de calcul :** (1) Standardisation des variables. (2) Ajout d'une pénalité $L_1$ (valeur absolue) à la fonction de coût. (3) Optimisation (par descente de coordonnées) pour annuler les coefficients non significatifs.
* **Paramètre appris :** Coefficients $\beta_j$ et constante $\beta_0$.
* **Hyperparamètre :** $\alpha$ (contrôle la force de la régularisation et le niveau de sélection de variables).

**Équation de la fonction de coût :**
$$J(\beta) = \frac{1}{2n} \sum_{i=1}^{n} \left( y_i - \beta_0 - \sum_{j=1}^{p} \beta_j x_{ij} \right)^2 + \alpha \sum_{j=1}^{p} |\beta_j|$$
* **$|\beta_j|$ :** Valeur absolue du coefficient de la variable $j$.
* **$\alpha$ :** Hyperparamètre de pénalité ($L_1$).

### C2 — Schéma original
```
                 | Coef Beta_2
                 |      / \ (Frontière géométrique en losange de la pénalité L1)
                 |     /   \
   --------------|----/-----*--- (Point d'intersection optimal souvent sur les axes)
                 |   /     /
                 |  /     /
                 | /     /
                 +------------ Coef Beta_1
```

### C3 — Contextes et choix
* **Contexte 1 (Gestion de la relation client) :** Déterminer les facteurs influençant la fidélité client parmi des centaines d'indicateurs comportementaux, en éliminant les critères redondants.
* **Contexte 2 (Finance / Économie) :** Sélectionner les indicateurs clés guidant l'inflation nationale parmi des dizaines de bases de données économiques.
* **Structure de données :** Tableau $n \times p$ avec potentiellement $p > n$ (haute dimension) et une cible continue.
* **Algorithme concurrent :** ElasticNet ou Régression Ridge.
* **Situation d'inadaptation :** En présence de variables très corrélées, le Lasso en choisit une arbitrairement et annule les autres, ce qui nuit à l'interprétabilité.
* **Métrique de performance :** R² et MAE.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Combien de variables sont annulées pour trois valeurs d’alpha ?*
Le Lasso réalise une sélection de variables intégrée. En faisant varier l'hyperparamètre $\alpha$, nous observons que plus $\alpha$ augmente, plus le nombre de variables dont le coefficient est strictement égal à zéro augmente.

````python
from sklearn.linear_model import Lasso

# Test de 3 valeurs d'alpha pour voir le nombre de variables annulées (coefficients nuls)
pourcentages_zeros = []
for a in [0.1, 1.0, 5.0]:
    modele_lasso = make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=15000)).fit(Xr_ent, yr_ent)
    coefs = modele_lasso[-1].coef_
    non_nuls = np.count_nonzero(coefs)
    nuls = len(coefs) - non_nuls
    mae = mean_absolute_error(yr_test, modele_lasso.predict(Xr_test))
    print(f"Alpha = {a:5} | Coefficients non nuls = {non_nuls} / 10 | Coefs annulés = {nuls} | MAE = {mae:.2f}")
````

## 4. Régression polynomiale

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, régression.
* **Entrées :** $p=10$ variables d'entrée transformées par combinaisons de degrés supérieurs.
* **Sortie :** $1$ variable continue.
* **Hypothèse clé :** La relation entre les variables prédictives et la variable cible n'est pas purement linéaire, mais présente des courbures ou des interactions multiplicatives.
* **Étapes de calcul :** (1) Génération des caractéristiques polynomiales d'ordre $d$. (2) Standardisation des nouvelles caractéristiques. (3) Entraînement d'un modèle linéaire (avec ou sans régularisation).
* **Paramètre appris :** Les coefficients $\beta_j$ associés à toutes les puissances et produits croisés des variables.
* **Hyperparamètre :** Le degré maximum $d$ des polynômes (`degree`).

**Équation (pour un modèle de degré 2 sur 2 variables) :**
$$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_1^2 + \beta_4 x_2^2 + \beta_5 x_1 x_2$$
* **$x_1, x_2$ :** Variables d'origine.
* **$x_1^2, x_2^2$ :** Termes quadratiques.
* **$x_1 x_2$ :** Terme d'interaction.
* **$\beta_j$ :** Coefficients appris pour chaque dimension étendue.

### C2 — Schéma original
```
     Y |               * (Points réels)
       |            *     
       |         *  ----
       |       *  /      \   <--- Ligne de régression polynomiale (courbe)
       |     *   /        \
       |   *    /          \
       +------------------------ X
```

### C3 — Contextes et choix
* **Contexte 1 (Agriculture / Énergie) :** Rendement d'une culture agricole en fonction de la température et de l'humidité (les extrêmes chauds ou froids réduisent la récolte, d'où l'aspect quadratique).
* **Contexte 2 (Physique industrielle) :** Frottements ou résistance d'une pièce mécanique selon la vitesse de rotation (croissance quadratique).
* **Structure de données :** Matrice d'entrée numérique continue à faible nombre de dimensions d'origine ($p$ petit pour éviter l'explosion combinatoire).
* **Algorithme concurrent :** SVR (RBF Kernel) ou Random Forest.
* **Situation d'inadaptation :** Très grand nombre de caractéristiques initiales (par exemple 100+), car le nombre de variables polynomiales de degré 2 ou 3 explose et entraîne un surapprentissage immédiat.
* **Métrique de performance :** $R^2$ de test.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Le degré 2 améliore-t-il le test ou seulement l’entraînement ?*
Nous allons tester les degrés 1, 2 et 3 pour comparer le score $R^2$ obtenu sur le jeu d'entraînement et sur le jeu de test afin de déterminer si l'ajout de termes non-linéaires généralise bien.

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

for deg in [1, 2, 3]:
    modele = make_pipeline(
        PolynomialFeatures(degree=deg, include_bias=False),
        StandardScaler(),
        Ridge(alpha=100.)
    ).fit(Xr_ent, yr_ent)
    r2_train = r2_score(yr_ent, modele.predict(Xr_ent))
    r2_test = r2_score(yr_test, modele.predict(Xr_test))
    print(f"Degré {deg} | R² Entraînement = {r2_train:.3f} | R² Test = {r2_test:.3f}")
````

## 5. Arbre de régression

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, régression.
* **Entrées :** $p=10$ variables quantitatives.
* **Sortie :** Valeur cible continue (valeur moyenne des observations de la feuille finale).
* **Hypothèse clé :** La relation peut être modélisée de manière non-linéaire par une série de découpages rectangulaires successifs de l'espace des caractéristiques.
* **Étapes de calcul :** (1) Recherche de la variable et du seuil qui minimisent la variance intra-nœud. (2) Séparation binaire de l'échantillon. (3) Répétition récursive jusqu'à atteindre un critère d'arrêt.
* **Paramètre appris :** La structure de l'arbre (les variables de coupure, les seuils, et les valeurs moyennes affectées aux feuilles).
* **Hyperparamètre :** `max_depth` (profondeur maximale de l'arbre).

**Équation du critère de division (variance / MSE) :**
$$H(Q) = \frac{1}{N_m} \sum_{i \in Q} (y_i - \bar{y}_m)^2$$
* **$Q$ :** Ensemble de données dans le nœud $m$.
* **$N_m$ :** Nombre d'échantillons au nœud $m$.
* **$\bar{y}_m$ :** Moyenne des cibles au nœud $m$.

### C2 — Schéma original
```
                    [ X_3 <= 0.5 ]  (Nœud racine)
                     /        \
                    /          \
              [ Feuille 1 ]   [ X_1 <= 12.4 ]
                               /         \
                         [ Feuille 2 ]  [ Feuille 3 ]
```

### C3 — Contextes et choix
* **Contexte 1 (Ressources Humaines) :** Prédire le salaire proposé à l'embauche en fonction de l'expérience, de l'âge et du niveau de diplôme (relations par paliers).
* **Contexte 2 (Immobilier) :** Estimation de la valeur d'une maison selon sa zone géographique et sa surface au sol.
* **Structure de données :** Mixte (numérique et catégorielle) sans besoin d'alignement d'échelle.
* **Algorithme concurrent :** Forêt aléatoire ou SVR.
* **Situation d'inadaptation :** Relations continues hautement linéaires ou données changeant de façon extrêmement continue (l'arbre produit des prédictions en escalier saccadées).
* **Métrique de performance :** MAE.

### C4 — Test de sensibilité et interprétation

## 7. Gradient Boosting — régression

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, régression.
* **Entrées :** $p=10$ variables quantitatives.
* **Sortie :** $1$ variable continue.
* **Hypothèse clé :** L'ajout séquentiel de modèles simples (arbres de faible profondeur) corrigeant pas à pas les erreurs résiduelles des étapes précédentes permet de construire un prédicteur puissant.
* **Étapes de calcul :** (1) Initialisation avec une prédiction constante. (2) Pour chaque itération, calcul des pseudo-résidus. (3) Ajustement d'un arbre sur ces résidus. (4) Mise à jour de la prédiction globale avec un taux d'apprentissage $\eta$.
* **Paramètre appris :** Les structures des arbres faibles successifs et leurs poids associés.
* **Hyperparamètre :** `max_iter` (nombre d'itérations / arbres) et `learning_rate` (vitesse d'apprentissage).

**Équation de récurrence :**
$$F_m(x) = F_{m-1}(x) + \eta \cdot h_m(x)$$
* **$F_m(x)$ :** Modèle cumulé à l'étape $m$.
* **$\eta$ :** Taux d'apprentissage (`learning_rate`).
* **$h_m(x)$ :** Arbre faible ajusté sur les résidus à l'étape $m$.

### C2 — Schéma original
```
    [ Modèle initial F_0 ] ===> Calcule l'erreur e_0
                                       |
                                       v
    [ Ajustement de l'arbre h_1 sur e_0 ] ===> Modèle F_1 = F_0 + eta*h_1
                                       |
                                       v
    [ Ajustement de l'arbre h_2 sur e_1 ] ===> Modèle F_2 = F_1 + eta*h_2
```

### C3 — Contextes et choix
* **Contexte 1 (Risque de crédit) :** Estimer la perte financière en cas de défaut d'un emprunteur.
* **Contexte 2 (Logistique) :** Prédiction précise du temps de livraison final pour optimiser les trajets des livreurs.
* **Structure de données :** Données tabulaires denses, de préférence avec de nombreuses observations.
* **Algorithme concurrent :** Forêt Aléatoire ou XGBoost.
* **Situation d'inadaptation :** Jeux de données extrêmement bruités ou très petits, où le Boosting a une tendance naturelle à surajuster le bruit.
* **Métrique de performance :** MAE et RMSE.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Que change max_iter sur le coût et le risque de surajustement ?*
Nous allons entraîner des modèles avec des valeurs croissantes de `max_iter` et évaluer le comportement de l'erreur sur le train et le test.

````python
from sklearn.ensemble import HistGradientBoostingRegressor

for iterations in [10, 50, 100, 300]:
    modele = HistGradientBoostingRegressor(max_iter=iterations, max_leaf_nodes=9, random_state=GRAINE).fit(Xr_ent, yr_ent)
    mae_train = mean_absolute_error(yr_ent, modele.predict(Xr_ent))
    mae_test = mean_absolute_error(yr_test, modele.predict(Xr_test))
    print(f"Itérations = {iterations:3} | MAE Train = {mae_train:.2f} | MAE Test = {mae_test:.2f}")
````

## 8. Régression logistique

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, classification.
* **Entrées :** $p=30$ variables réelles (caractéristiques géométriques des cellules) du jeu *Breast Cancer*.
* **Sortie :** Classe binaire $y \in \{0, 1\}$ (tumeur maligne ou bénigne).
* **Hypothèse clé :** Le log-odds (logarithme du ratio des probabilités) des classes est une combinaison linéaire des variables d'entrée.
* **Étapes de calcul :** (1) Standardisation des variables. (2) Calcul de la combinaison linéaire $z$. (3) Passage par la fonction logistique (sigmoïde) pour obtenir la probabilité. (4) Maximisation de la vraisemblance (cross-entropy loss).
* **Paramètre appris :** Les poids $\beta_j$ et le biais $\beta_0$.
* **Hyperparamètre :** L'inverse de la force de régularisation $C$ (plus $C$ est petit, plus la régularisation est forte).

**Équation de prédiction (Sigmoïde) :**
$$p(y=1|x) = \sigma(\beta_0 + \sum_{j=1}^{p} \beta_j x_j) = \frac{1}{1 + e^{-(\beta_0 + \sum_{j=1}^{p} \beta_j x_j)}}$$
* **$\sigma(z)$ :** Fonction sigmoïde projetant sur $]0, 1[$.
* **$x_j$ :** Variable d'entrée $j$.
* **$\beta_j$ :** Coefficient de la variable $j$.

### C2 — Schéma original
```
   Probabilité p
        1.0 |                . - - - - - *
            |            .
        0.5 | - - - - * - - - - - - - (Seuil de décision à z = 0)
            |       .
        0.0 | * - -
            +---------------------------- combinatoire z = beta*x
```

### C3 — Contextes et choix
* **Contexte 1 (Marketing / Vente) :** Prédire si un internaute va souscrire à un abonnement premium après avoir reçu une offre promotionnelle.
* **Contexte 2 (Banque / Risque) :** Score d'octroi de crédit pour évaluer si un client va faire défaut.
* **Structure de données :** Tableau $n \times p$ de caractéristiques continues ou binaires et une étiquette de classe binaire.
* **Algorithme concurrent :** SVM linéaire ou Naive Bayes.
* **Situation d'inadaptation :** Relations fortement non-linéaires complexes ou interactions d'ordre élevé non modélisées explicitement.
* **Métrique de performance :** F1-score (car équilibre la précision et le rappel sur la détection des tumeurs malignes).

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Quel effet de C sur les probabilités et la frontière ?*
Nous allons entraîner des modèles avec des valeurs croissantes de $C$ pour voir comment la régularisation modifie le score F1 final sur le test.

````python
from sklearn.linear_model import LogisticRegression

for c_val in [0.001, 0.05, 0.5, 10.0]:
    modele = make_pipeline(StandardScaler(), LogisticRegression(C=c_val, max_iter=2000, random_state=GRAINE)).fit(Xc_ent, yc_ent)
    f1_val = f1_score(yc_test, modele.predict(Xc_test))
    print(f"C = {c_val:7.3f} | F1-Score Test = {f1_val:.3f}")
````

* **Hypothèse clé :** Les observations proches dans l'espace des caractéristiques appartiennent généralement à la même classe.
* **Étapes de calcul :** (1) Standardisation des variables. (2) Calcul de la distance (souvent euclidienne) entre le point à prédire et tous les points d'entraînement. (3) Sélection des $k$ voisins les plus proches. (4) Vote majoritaire.

**Équation de distance (Minkowski) :**
$$d(x, z) = \left( \sum_{i=1}^{p} |x_i - z_i|^r \right)^{1/r}$$
* **$x, z$ :** Deux observations.
* **$r$ :** Paramètre de distance ($r=2$ pour la distance euclidienne).

                 \  
                  \  (k=3 voisins proches)

                  /
                 /
                * (Classe B) => Classification finale: Classe B (vote majoritaire)
```

* **Structure de données :** Tableau numérique compact exempt de valeurs manquantes.
* **Algorithme concurrent :** SVM à noyau RBF.

* **Sortie :** Classe binaire $y \in \{0, 1\}$.
* **Hypothèse clé :** Hypothèse d'indépendance conditionnelle forte : toutes les caractéristiques d'entrée sont indépendantes entre elles sachant la valeur de la classe cible.
* **Étapes de calcul :** (1) Estimation de la moyenne et de la variance de chaque variable pour chaque classe. (2) Calcul des densités de probabilité gaussiennes conditionnelles. (3) Calcul de la probabilité a posteriori par théorème de Bayes. (4) Sélection de la classe de probabilité maximale.
* **Paramètres appris :** Les moyennes $\mu_{jc}$ et variances $\sigma^2_{jc}$ de chaque variable pour chaque classe.
* **Hyperparamètre :** Le lissage de variance `var_smoothing` (pour éviter de diviser par des valeurs proches de zéro).

**Équation clé (Théorème de Bayes avec indépendance) :**
$$P(Y=c|X) \propto P(Y=c) \prod_{j=1}^{p} P(X_j|Y=c)$$
* **$P(Y=c)$ :** Probabilité a priori de la classe $c$.
* **$P(X_j|Y=c)$ :** Densité gaussienne de la caractéristique $j$.

### C2 — Schéma original
```
                    [ Variable Cible Y ]
                     /       |        \
                    v        v         v
                [ Var X_1 ] [ Var X_2 ] [ Var X_3 ]

```

* **Contexte 2 (Sécurité) :** Détection d'emails frauduleux ou de spams bancaires.
* **Structure de données :** Tableau $n \times p$ de caractéristiques textuelles (sac de mots) ou continues.
* **Algorithme concurrent :** Régression Logistique.
* **Situation d'inadaptation :** En présence de variables d'entrées fortement corrélées (colinéarité), car l'hypothèse d'indépendance conditionnelle est violée, ce qui déforme les probabilités a posteriori.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, classification.
* **Entrées :** $p=30$ variables réelles (base *Breast Cancer*).
* **Sortie :** Classe binaire $y \in \{-1, 1\}$ ou $\{0, 1\}$.
* **Hypothèse clé :** Il existe un hyperplan optimal séparant les classes avec une marge maximale, éventuellement dans un espace de projection de plus haute dimension via l'astuce du noyau (Kernel Trick).
* **Étapes de calcul :** (1) Standardisation des variables. (2) Calcul de la matrice de noyau. (3) Résolution du problème d'optimisation quadratique dual sous contraintes. (4) Détermination des vecteurs supports.
* **Paramètre appris :** Les multiplicateurs de Lagrange $\alpha_i$ et l'ordonnée à l'origine $b$.
* **Hyperparamètre :** Le paramètre de régularisation $C$ et le type de noyau (`kernel`).

**Équation de l'hyperplan optimal (Dual) :**
$$f(x) = \text{sign} \left( \sum_{i=1}^{n_s} \alpha_i y_i K(x_i, x) + b \right)$$
* **$n_s$ :** Nombre de vecteurs supports.
* **$K(x_i, x)$ :** Fonction de noyau (linéaire, RBF, polynomial, etc.).
* **$y_i$ :** Étiquette de classe de l'échantillon $i$.
* **$\alpha_i$ :** Coefficients de Lagrange appris.

### C2 — Schéma original
```
          Classe +1  *      *  (Vecteur support)
                       \    |
                        \   | Marge (1 / ||w||)
     --------------------\--|----------------- Hyperplan séparateur optimal
                          \ |
                           \|
          Classe -1  o      o  (Vecteur support)
```

### C3 — Contextes et choix
* **Contexte 1 (Bioinformatique) :** Classification de patients sains ou atteints d'une maladie rare à partir de profils d'expression génomique (haute dimension $p \gg n$).
* **Contexte 2 (Vision par ordinateur) :** Reconnaissance optique de caractères (chiffres écrits à la main).
* **Structure de données :** Tableau numérique de variables standardisées, sans valeurs manquantes.
* **Algorithme concurrent :** Forêt Aléatoire ou MLP (Réseau de neurones).
* **Situation d'inadaptation :** Très grand jeu de données ($n > 100\ 000$ d'observations), car le coût de calcul de la matrice de noyau devient prohibitif ($O(n^3)$).
* **Métrique de performance :** F1-score.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Comparez noyau linéaire et RBF, et justifiez le choix.*
Nous allons tester le noyau linéaire face au noyau RBF pour évaluer si la frontière de décision nécessite une transformation non-linéaire complexe pour optimiser le score F1.

## 14. Gradient Boosting — classification

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, classification par boosting.
* **Entrées :** $p=30$ variables de la base *Breast Cancer*.
* **Sortie :** Classe binaire $y \in \{0, 1\}$.
* **Hypothèse clé :** La combinaison séquentielle d'arbres de décision simples (qui prédisent la dérivée de la perte logistique) converge vers un classifieur puissant.
* **Étapes de calcul :** (1) Initialisation avec une prédiction constante (log-odds moyen). (2) Calcul des résidus d'entropie croisée. (3) Ajustement d'un arbre sur ces résidus. (4) Ajout pondéré au modèle cumulatif.
* **Paramètre appris :** La structure de tous les arbres faibles entraînés sur les résidus.
* **Hyperparamètre :** Nombre maximal d'itérations `max_iter` et le taux d'apprentissage `learning_rate`.

**Équation de perte logistique (Cross-Entropy loss) minimisée :**
$$L(y, p) = - \sum_{i=1}^{n} \left[ y_i \log(p_i) + (1 - y_i) \log(1 - p_i) \right]$$
* **$y_i$ :** Vraie classe de l'observation $i$.
* **$p_i$ :** Probabilité prédite par l'ensemble à l'étape actuelle.

### C2 — Schéma original
```
     Étape 1 : Arbre F_0 (prédiction de base) ===> Erreur résiduelle 1
                                                         |
                                                         v
     Étape 2 : Arbre h_1 (ajusté sur résidus 1) ===> F_1 = F_0 + eta * h_1 ===> Erreur résiduelle 2
```

### C3 — Contextes et choix

            [ Classifieur faible 2 ]

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé, classification.
* **Entrées :** $p=30$ variables réelles.
* **Sortie :** Classe binaire $y \in \{0, 1\}$.
* **Hypothèse clé :** L'ajout d'une régularisation formelle directement dans la fonction objective du boosting d'arbres permet de maximiser la vitesse d'entraînement tout en évitant le surapprentissage.
* **Étapes de calcul :** (1) Calcul des gradients du premier et second ordre ($g_i$ et $h_i$) pour chaque exemple. (2) Construction d'arbres en cherchant la division qui maximise le gain régularisé. (3) Application d'un élagage basé sur le paramètre $\gamma$.
* **Paramètre appris :** Les coefficients aux feuilles et la structure de l'ensemble d'arbres.
* **Hyperparamètre :** La profondeur maximale des arbres `max_depth`.

**Équation de la fonction objective régularisée :**
$$\mathcal{L}^{(t)} \approx \sum_{i=1}^{n} \left[ g_i f_t(x_i) + \frac{1}{2} h_i f_t^2(x_i) \right] + \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2$$
* **$g_i, h_i$ :** Gradients de premier et second ordre de la perte.
* **$T$ :** Nombre de feuilles de l'arbre.
* **$w_j$ :** Poids de la feuille $j$.

### C2 — Schéma original
```
                     [ Résidus courants ]
                              |
         (Calcul des Gradients g_i et Hessiens h_i)
                              v
         [ Splitting basé sur le gain de Bellman régularisé ]
                              |
                              v
                [ Arbre régularisé optimal ]
```

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, clustering.
* **Entrées :** $p=13$ variables d'entrée (base *Wine* standardisée).
* **Sortie :** Étiquettes de groupe $k \in \{0, 1, \dots, K-1\}$.
* **Hypothèse clé :** Les grappes sont sphériques, de tailles similaires et bien séparées.
* **Étapes de calcul :** (1) Initialisation aléatoire ou via k-means++ de $K$ centroïdes. (2) Assignation de chaque observation au centroïde le plus proche (distance euclidienne). (3) Recalcul des centroïdes comme moyenne des points de chaque groupe. (4) Répétition jusqu'à convergence.
* **Paramètre appris :** Les coordonnées des centroïdes finaux.
* **Hyperparamètre :** Le nombre de clusters $K$ (`n_clusters`).

**Équation de la fonction objectif (Inertie intra-classe) :**
$$J = \sum_{k=1}^{K} \sum_{i \in S_k} \|x_i - \mu_k\|^2$$
* **$S_k$ :** Ensemble des points du cluster $k$.
* **$\mu_k$ :** Centroïde du cluster $k$.
* **$\|x_i - \mu_k\|^2$ :** Carré de la distance euclidienne.

### C2 — Schéma original
```
    [ Initialisation : Centroïdes placés au hasard ]
                           |
                           v
    [ Étape Assignation : Points rattachés au centroïde le plus proche ]
                           |
                           v
    [ Étape Mise à jour : Centroïdes déplacés vers le centre de gravité des points ]
                           |
    (Convergence) <--------+
```

### C3 — Contextes et choix

* **Contexte 2 (Sécurité) :** Regroupement de profils de connexion réseau pour détecter des comportements de bande passante atypiques.
* **Structure de données :** Tableau numérique d'observations standardisées et quantitatives.
* **Algorithme concurrent :** Mélanges gaussiens (GMM) ou DBSCAN.
* **Situation d'inadaptation :** Clusters de formes géométriques complexes non convexes (ex: croissants, cercles concentriques) ou tailles extrêmement disproportionnées.
* **Métrique de performance :** Score de Silhouette et coefficient d'Inertie.

* **Entrées :** $p=13$ variables d'entrée.

* **Étapes de calcul :** (1) Initialisation : chaque point forme son propre groupe. (2) Calcul de la matrice de distances entre tous les groupes. (3) Fusion des deux groupes les plus proches selon le critère de liaison (linkage). (4) Répétition jusqu'à obtenir un seul grand groupe.

**Équation de la liaison de Ward (augmentation de variance) :**

* **$n_A, n_B$ :** Nombre d'éléments dans les groupes $A$ et $B$.
* **$\mu_A, \mu_B$ :** Centroïdes respectifs des groupes.

### C2 — Schéma original

            [ A ]    [ B ]    [ C ]    [ D ]  (Observations individuelles)
              \       /         |        |
               [ A-B ]          |        |
                  \            /         |
                   [  A-B-C  ]           |
                         \              /
                          [  A-B-C-D  ]   (Groupe racine unique)
```

* **Situation d'inadaptation :** Très grand nombre d'observations ($n > 20\ 000$), car la complexité en temps de $O(n^3)$ et en espace de $O(n^2)$ sature rapidement les ressources.

## 19. DBSCAN

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, clustering basé sur la densité.
* **Entrées :** $p=13$ variables d'entrée.
* **Sortie :** Étiquettes de groupes et identification des points de bruit (étiquette $-1$).
* **Hypothèse clé :** Les clusters sont formés de zones denses séparées par des régions de faible densité, sans forme géométrique prédéfinie.
* **Étapes de calcul :** (1) Pour chaque point, calcul du nombre d'échantillons dans son voisinage de rayon $\epsilon$. (2) Si ce nombre est supérieur à `min_samples`, le point devient un point cœur (Core Point). (3) Propagation récursive pour regrouper tous les points cœurs connectés. (4) Les points non atteints sont déclarés comme du bruit.
* **Paramètre appris :** Aucun (algorithme à base de densité, pas de centroïde formel).
* **Hyperparamètre :** Le rayon de voisinage $\epsilon$ (`eps`) et le nombre minimal de points requis `min_samples`.

**Équation formelle du voisinage $\epsilon$ :**
$$N_{\epsilon}(x) = \{ z \in D \mid d(x, z) \le \epsilon \}$$
* **$d(x, z)$ :** Distance (généralement euclidienne).
* **$\epsilon$ :** Seuil de distance d'inclusion.

### C2 — Schéma original
```
                 * (Point Cœur : >= MinSamples dans son rayon)
                / \
               *---* (Points frontières : proches d'un cœur mais pas assez d'échantillons propres)

                                  o (Point Bruit : isolé dans l'espace)
```

### C3 — Contextes et choix
* **Contexte 1 (Géolocalisation) :** Regroupement spatial de points d'intérêt GPS (par exemple, identification de zones d'activité commerciale forte).
* **Contexte 2 (Sécurité) :** Détection de fraude à la carte bancaire en séparant les comportements normaux (clusters denses) des anomalies (bruit).
* **Structure de données :** Tableau numérique continu continu multi-dimensionnel.
* **Algorithme concurrent :** K-means ou OPTICS.
* **Situation d'inadaptation :** En présence de densités de grappes extrêmement variables, car un unique paramètre $\epsilon$ ne peut pas capturer simultanément des groupes de densités très différentes.
* **Métrique de performance :** Nombre de groupes détectés et pourcentage de points considérés comme du bruit.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Expliquez pourquoi eps modifie le nombre de points bruit.*
L'hyperparamètre `eps` ($\epsilon$) contrôle directement la taille du voisinage exploré. Si $\epsilon$ est trop petit, très peu de points parviennent à satisfaire le critère de densité minimale, ce qui augmente mécaniquement le nombre de points exclus (considérés comme du bruit). Nous allons le tester pour deux valeurs de rayon.

````python
from sklearn.cluster import DBSCAN

for epsilon in [1.8, 2.5, 3.2]:
    etiquettes = DBSCAN(eps=epsilon, min_samples=4).fit_predict(X_clu)
    groupes = len(set(etiquettes) - {-1})
    bruit = np.sum(etiquettes == -1)
    print(f"Eps = {epsilon:.1f} | Groupes détectés hors bruit = {groupes} | Points considérés comme bruit = {bruit} / {len(X_clu)}")
````

## 20. Mélanges gaussiens

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, clustering probabiliste.
* **Entrées :** $p=13$ variables d'entrée.
* **Sortie :** Vecteur de probabilités d'appartenance à chaque composante (soft clustering) et étiquettes de groupes.
* **Hypothèse clé :** Toutes les données proviennent d'une superposition de plusieurs distributions gaussiennes multidimensionnelles indépendantes ayant chacune leurs propres moyennes et matrices de covariance.
* **Étapes de calcul :** Algorithme Espérance-Maximisation (EM) : (1) Étape E : calcul de la probabilité a posteriori que chaque point appartienne à chaque composante. (2) Étape M : mise à jour des paramètres (moyennes, covariances et poids des mélanges) pour maximiser la vraisemblance.
* **Paramètres appris :** Les moyennes $\mu_k$, les matrices de covariance $\Sigma_k$ et les poids du mélange $\pi_k$.
* **Hyperparamètre :** Le type de matrice de covariance `covariance_type` ('full', 'tied', 'diag', 'spherical').

**Équation de densité du mélange :**
$$p(x) = \sum_{k=1}^{K} \pi_k \mathcal{N}(x \mid \mu_k, \Sigma_k)$$
* **$\pi_k$ :** Poids de la composante $k$.
* **$\mathcal{N}$ :** Loi normale multidimensionnelle.

### C2 — Schéma original
```
                  Composante 1      Composante 2
                     (  *  )           (  o  )
                    ( * * * )         ( o o o )
                   (  * *  )           (  o  )
        (Frontières de décision douces / ellipsoïdales orientables)
```

* **Structure de données :** Tableau numérique continu standardisé.
* **Algorithme concurrent :** K-means.
* **Situation d'inadaptation :** Un volume de données très restreint face à un grand nombre de dimensions, ce qui conduit à l'incapacité d'estimer correctement les matrices de covariance complètes (singularité).
* **Métrique de performance :** Score de Silhouette.

**Variante C — question spécifique :** *Comparez covariance_type = full et diag.*
Nous allons tester l'effet du choix de la structure de la matrice de covariance (`diag` vs `full`) sur le score de silhouette afin de voir si la simplification géométrique influence la qualité des grappes obtenues.

## 21. ACP

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, réduction de dimension linéaire.
* **Entrées :** $p=13$ variables d'entrée (base *Wine*).
* **Sortie :** $k$ composantes principales projetées d'une variance maximale ($k < p$).
* **Hypothèse clé :** Les directions de plus grande variance contiennent l'information la plus utile de la structure globale des données.
* **Étapes de calcul :** (1) Centrage et réduction des variables d'entrée. (2) Calcul de la matrice de covariance. (3) Calcul des valeurs propres et vecteurs propres de cette matrice. (4) Projection orthogonale des données sur les vecteurs associés aux plus grandes valeurs propres.
* **Paramètre appris :** Les vecteurs propres (chargements / loadings) et valeurs propres.
* **Hyperparamètre :** Le nombre de composantes principales `n_components`.

**Équation de projection (Composante principale $z_i$) :**
$$z_i = w_i^T x = \sum_{j=1}^{p} w_{ij} x_j$$
* **$x$ :** Vecteur d'entrée d'origine.
* **$w_i$ :** Vecteur propre normé ($w_i^T w_i = 1$) maximisant la variance.

### C2 — Schéma original
```
          X_2 ^
              |         *  (Points d'origine)
              |      *   / (Axe de plus grande variance : PC1)
              |    *    /
              |  *     /   _ _ (Axe orthogonal : PC2)
              |*      /
              +--------------------> X_1
```

### C3 — Contextes et choix
* **Contexte 1 (Finance) :** Modélisation de la courbe des taux d'intérêt nationaux à l'aide de dizaines d'échéances d'obligations d'État.
* **Contexte 2 (Marketing / Enquêtes) :** Compression d'un questionnaire de satisfaction client de 100 questions pour dégager 3 dimensions d'opinion globale.
* **Structure de données :** Tableau numérique continu continu.
* **Algorithme concurrent :** t-SNE ou UMAP (si les relations sont fortement non-linéaires).
* **Situation d'inadaptation :** Relations entre caractéristiques de nature purement non-linéaire (ex : une structure toroïdale ou sinusoïdale complexe).
* **Métrique de performance :** Ratio de variance expliquée cumulée.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Combien de composantes faut-il pour 80 % de variance ?*
Nous allons projeter les données et calculer la somme cumulée des variances expliquées par chaque axe pour déterminer précisément le nombre minimal d'axes requis.

````python
from sklearn.decomposition import PCA
import numpy as np

modele = PCA(random_state=GRAINE).fit(X_clu)
variance_cumulee = np.cumsum(modele.explained_variance_ratio_)
for idx, var in enumerate(variance_cumulee):
    print(f"{idx+1} composante(s) | Variance expliquée cumulée = {var:.3%}")

# Extraction exacte
nb_recomp = np.argmax(variance_cumulee >= 0.80) + 1
print(f"\n===> Il faut donc précisément {nb_recomp} composantes pour capturer au moins 80% de la variance.")
````

## 22. t-SNE

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, réduction de dimension non-linéaire (visualisation).
* **Entrées :** $p=13$ variables d'entrée.
* **Sortie :** Coordonnées bidimensionnelles ou tridimensionnelles (généralement 2D).
* **Hypothèse clé :** La préservation des voisinages locaux et des distances à petite échelle permet d'obtenir une représentation lisible des regroupements internes complexes.
* **Étapes de calcul :** (1) Calcul des probabilités de voisinage conditionnelles gaussiennes dans l'espace d'origine. (2) Définition des probabilités de voisinage via une loi de Student à 1 degré de liberté dans l'espace réduit (pour éviter l'écrasement). (3) Minimisation de la divergence de Kullback-Leibler par descente de gradient.
* **Paramètre appris :** Les coordonnées réduites finales des observations d'origine.
* **Hyperparamètre :** La perplexité `perplexity` (qui estime grossièrement le nombre effectif de voisins).

**Équation de la divergence de Kullback-Leibler (Fonction objectif) :**
$$KL(P || Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$
* **$p_{j|i}$ :** Probabilité de similarité locale dans l'espace d'origine.
* **$q_{j|i}$ :** Probabilité de similarité locale dans l'espace de projection.

### C2 — Schéma original
```
     [ Espace réel d'origine ] ===> Probabilités de voisinage gaussiennes P
                                          |
                                          | (Préservation de la structure locale)
                                          v
     [ Espace de projection 2D ] ===> Probabilités t-Student Q, minimisation KL
```

### C3 — Contextes et choix
* **Contexte 1 (Bioinformatique) :** Visualisation de types de cellules de tissus sains ou cancéreux à l'aide de mesures d'ARN de haute dimension (single-cell RNA-seq).
* **Contexte 2 (Analyse web) :** Représentation des profils utilisateurs issus de graphes de navigation sur les réseaux sociaux.
* **Structure de données :** Tableau numérique continu multi-dimensionnel.
* **Algorithme concurrent :** UMAP (souvent plus rapide et préservant mieux les structures globales).
* **Situation d'inadaptation :** Si l'objectif principal est de reconstruire les données d'origine (t-SNE ne définit pas de fonction de projection réversible) ou d'évaluer les distances à grande échelle.
* **Métrique de performance :** Score de divergence finale de Kullback-Leibler.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *La structure change-t-elle quand perplexity passe de 12 à 30 ?*

## 23. UMAP

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, réduction de dimension non-linéaire.
* **Entrées :** $p=13$ variables d'entrée.
* **Sortie :** Coordonnées de projection bidimensionnelles.
* **Hypothèse clé :** Les données résident sur une variété topologique Riemannienne locale localement connexe.
* **Étapes de calcul :** (1) Construction d'un graphe de voisinage pondéré flou basé sur des métriques de distance locales. (2) Établissement d'une représentation topologique équivalente de basse dimension. (3) Optimisation de la disposition de basse dimension par l'entropie croisée floue.
* **Paramètre appris :** La structure des coordonnées de projection finales.
* **Hyperparamètre :** Nombre de voisins `n_neighbors` et distance minimale `min_dist`.

**Équation de l'entropie croisée floue (Fonction de perte) :**
$$\Phi(X, Y) = \sum_{e \in E} \left[ \mu_e \log \nu_e + (1 - \mu_e) \log (1 - \nu_e) \right]$$
* **$\mu_e$ :** Poids de l'arête $e$ dans le graphe d'origine.
* **$\nu_e$ :** Poids de l'arête $e$ dans le graphe de projection.

### C2 — Schéma original
```
     [ Graphe flou d'origine ] === (Préservation locale ET globale) ===> [ Graphe projeté ]
```

### C3 — Contextes et choix
* **Contexte 1 (Génomique) :** Cartographie de l'évolution de populations de cellules biologiques à travers le temps.
* **Contexte 2 (Traitement du signal) :** Visualisation interactive d'un corpus massif de voix numérisées transcrites.
* **Structure de données :** Tableau continu de haute dimension.
* **Algorithme concurrent :** t-SNE.
* **Situation d'inadaptation :** Scénario requérant une interprétabilité simple et directe de chaque dimension de sortie sous forme linéaire.
* **Métrique de performance :** Score de silhouette obtenu sur la projection 2D.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Que provoque n_neighbors = 8 puis 30 ?*
Le nombre de voisins `n_neighbors` contrôle la mise au point locale ou globale de l'algorithme. Nous allons le tester pour deux valeurs distinctes.

````python
import umap
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for idx, neighbors in enumerate([8, 30]):
    projection = umap.UMAP(n_components=2, n_neighbors=neighbors, min_dist=.25, random_state=GRAINE).fit_transform(X_clu)
    axes[idx].scatter(projection[:, 0], projection[:, 1], c=jeu_non_supervise.target, cmap='plasma', s=18)
    axes[idx].set_title(f'UMAP (n_neighbors={neighbors})')
plt.show()
````

## 24. Isolation Forest

### C1 — Présentation
* **Catégorie :** Apprentissage non supervisé, détection d'anomalies.
* **Entrées :** $p=13$ variables d'entrée.
* **Sortie :** Étiquette d'anomalie $y \in \{-1, 1\}$ ($-1$ pour les points anormaux, $1$ pour les points normaux).
* **Hypothèse clé :** Les anomalies sont rares, différentes et donc plus faciles à isoler par divisions binaires aléatoires que les points sains.
* **Étapes de calcul :** (1) Sélection d'une caractéristique aléatoire et d'un seuil de coupure aléatoire. (2) Division récursive de l'espace pour chaque arbre. (3) Calcul de la longueur moyenne du chemin pour chaque échantillon. (4) Conversion de cette longueur en un score d'anomalie.
* **Paramètres appris :** La structure de l'ensemble d'arbres d'isolation binaires.
* **Hyperparamètre :** Le taux de contamination estimé `contamination`.

**Équation du score d'anomalie :**
$$s(x, n) = 2^{-\frac{E(h(x))}{c(n)}}$$
* **$E(h(x))$ :** Longueur moyenne du chemin de l'observation $x$ à travers les arbres.
* **$c(n)$ :** Longueur moyenne de chemin d'un arbre binaire de recherche infructueux à $n$ nœuds.

### C2 — Schéma original
```
         Donnée normale (Isolée en de nombreuses étapes) :
         [ Racine ] -> [ Droite ] -> [ Gauche ] -> [ Droite ] -> [ Feuille ] (Chemin long)

         Donnée anormale (Isolée très rapidement) :
         [ Racine ] -> [ Gauche ] (Isolé instantanément - Chemin court)
```

### C3 — Contextes et choix
* **Contexte 1 (Cybersécurité) :** Détection d'intrusions sur des serveurs informatiques en identifiant les requêtes atypiques.
* **Contexte 2 (Industrie) :** Détection de pannes d'usines à partir de données temporelles issues de capteurs connectés IoT.
* **Structure de données :** Tableau de caractéristiques denses ou creuses.
* **Algorithme concurrent :** One-Class SVM ou Local Outlier Factor (LOF).
* **Situation d'inadaptation :** Si le pourcentage d'anomalies réelles dans le jeu de données d'entraînement dépasse $50\%$, violant l'hypothèse de rareté.
* **Métrique de performance :** Nombre de détections aberrantes.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Comparez contamination = 0.03 et 0.10.*
Nous allons modifier l'hyperparamètre `contamination` pour observer son influence directe sur le volume d'échantillons classés hors-norme.

````python
from sklearn.ensemble import IsolationForest

for cont in [0.03, 0.10]:
    modele = IsolationForest(contamination=cont, n_estimators=120, random_state=GRAINE).fit(X_clu)
    etiquettes = modele.predict(X_clu)
    nb_anomalies = (etiquettes == -1).sum()
    print(f"Contamination = {cont:.2f} | Observations atypiques détectées = {nb_anomalies} / {len(X_clu)}")
````

## 25. Bandits manchots

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement, optimisation séquentielle.
* **Entrées :** Pas de caractéristiques d'état (contexte vide), $K=4$ actions (bras du bandit).
* **Sortie :** Choix d'un bras et observation de la récompense associée $r_t$.
* **Hypothèse clé :** Chaque bras produit des récompenses selon une distribution de probabilité inconnue mais stationnaire.
* **Étapes de calcul :** (1) Sélection du bras $a_t$ à tester. (2) Observation de la récompense $r_t$. (3) Mise à jour de la moyenne empirique de gain de ce bras.
* **Paramètre appris :** Les récompenses moyennes estimées de chaque action $Q(a)$.
* **Hyperparamètre :** Nombre total d'itérations (tirages).

**Équation de mise à jour empirique :**
$$Q_{t+1}(a) = Q_t(a) + \frac{1}{N_t(a)} \left[ R_t - Q_t(a) \right]$$
* **$Q_t(a)$ :** Valeur estimée du bras $a$ à l'étape $t$.
* **$N_t(a)$ :** Nombre d'essais du bras $a$.
* **$R_t$ :** Récompense observée.

### C2 — Schéma original
```
                    [ Agent Bandit ]
                       /  |  |  \
                      v   v  v   v
                    [ B1 B2 B3 B4 ]  (Bras)
                         |
                         v
                  [ Récompense R ] ===> Mise à jour de la valeur moyenne estimée
```

### C3 — Contextes et choix
* **Contexte 1 (E-commerce) :** Test A/B dynamique pour identifier le bandeau promotionnel le plus cliqué par les visiteurs.
* **Contexte 2 (Finance / Portefeuille) :** Allocation de budget marketing entre 4 plateformes d'acquisition d'audience.
* **Structure de données :** Historique séquentiel (bras choisi, récompense reçue).
* **Algorithme concurrent :** Epsilon-Greedy ou Upper Confidence Bound (UCB).
* **Situation d'inadaptation :** Si le choix optimal dépend d'informations contextuelles variables (ex: météo ou profil démographique précis du client), car les bandits simples ignorent le contexte.
* **Métrique de performance :** Récompense cumulée (ou regret cumulé).

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Quel bras obtient la meilleure moyenne après 800 tirages ?*
Nous allons lancer la simulation pour déterminer quel bras a obtenu le taux moyen de récompense empirique le plus élevé.

````python
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
for etape in range(800):
    bras = etape % 4
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
moyennes = recompenses / nombre_tirages
meilleur_bras = np.argmax(moyennes)
print('Récompenses moyennes estimées :', np.round(moyennes, 3))
print(f"Le meilleur bras empirique après 800 tirages est le Bras {meilleur_bras} (index) avec {moyennes[meilleur_bras]:.3f} de moyenne.")
````

## 26. Epsilon-greedy

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement (compromis exploration / exploitation).
* **Entrées :** $K=4$ actions.
* **Sortie :** Index de l'action choisie.
* **Hypothèse clé :** L'exploration uniforme à une fréquence $\epsilon$ permet d'éviter de converger prématurément vers un sous-optimum local.
* **Étapes de calcul :** (1) Génération d'un nombre aléatoire $u \sim U(0, 1)$. (2) Si $u < \epsilon$, choix d'une action au hasard (exploration). (3) Sinon, choix de l'action maximisant les gains moyens estimés (exploitation).
* **Paramètres appris :** Valeurs d'action estimées $Q(a)$.
* **Hyperparamètre :** Taux d'exploration $\epsilon$.

**Équation de sélection de politique :**
$$a_t = \begin{cases} \text{action aléatoire} & \text{avec probabilité } \epsilon \\ \text{argmax}_{a} Q_t(a) & \text{avec probabilité } 1-\epsilon \end{cases}$$

### C2 — Schéma original
```
                    [ Choix de l'action ]
                     /                 \
         (u < epsilon) /                   \ (u >= epsilon)
                     v                     v
            [ Exploration : ]            [ Exploitation : ]
          (Sélection aléatoire)       (Sélection de l'Argmax Q(a))
```

### C3 — Contextes et choix
* **Contexte 1 (Routage Réseau) :** Test de nouveaux chemins réseau alternatifs tout en exploitant majoritairement la ligne la plus rapide.
* **Contexte 2 (Marketing) :** Recommandation adaptative de produits sur une plateforme web.
* **Structure de données :** Historique séquentiel de couples d'action-récompense.
* **Algorithme concurrent :** Thompson Sampling.
* **Situation d'inadaptation :** Environnement hautement instable où la distribution de récompense change à chaque instant.
* **Métrique de performance :** Valeur moyenne cumulée.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Quel effet de epsilon sur la concentration des choix ?*
Nous allons tester $\epsilon = 0.05$ (faible exploration) et $\epsilon = 0.40$ (forte exploration) pour mesurer comment l'allocation des choix se répartit sur les bras.

````python
for eps in [0.05, 0.40]:
    rng = np.random.default_rng(GRAINE)
    recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
    for etape in range(1500):
        if rng.random() < eps or not np.all(nombre_tirages):
            bras = int(rng.integers(4))
        else:
            bras = int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
        gain = float(rng.random() < probabilites_simulees[bras])
        recompenses[bras] += gain; nombre_tirages[bras] += 1
    print(f"Epsilon = {eps:.2f} | Sélection des bras : {nombre_tirages.astype(int)} | Valeurs estimées : {np.round(recompenses / np.maximum(nombre_tirages, 1), 3)}")
````

## 27. Q-learning

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement sans modèle (off-policy).
* **Entrées :** État courant $s_t$.
* **Sortie :** Action optimale $a_t$.
* **Hypothèse clé :** L'estimation de la fonction de valeur d'action optimale $Q^*(s, a)$ s'effectue de manière itérative via l'équation d'optimalité de Bellman, indépendamment de la politique de décision suivie.
* **Étapes de calcul :** (1) Observation de l'état $s_t$ et choix de l'action $a_t$ (via politique $\epsilon$-greedy). (2) Réception de la récompense $r_{t}$ et transition vers $s_{t+1}$. (3) Mise à jour de la table $Q$.
* **Paramètres appris :** La table de valeurs d'actions $Q(s, a)$.
* **Hyperparamètre :** Taux d'apprentissage $\alpha$ (learning rate) et facteur d'actualisation $\gamma$ (discount factor).

**Équation de mise à jour temporelle :**
$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
* **$\max_{a'} Q(s', a')$ :** Maximisation de l'estimation de valeur de l'état suivant (caractère Off-Policy).

### C2 — Schéma original
```
    [ État s ] ---> Action a (Choix e-greedy) ---> Récompense R et État s'
                                                          |
         [ Mise à jour Off-Policy avec l'estimation max_a' Q(s', a') ] <---+
```

### C3 — Contextes et choix
* **Contexte 1 (Jeux Vidéo / Robotique) :** Apprentissage de la trajectoire optimale d'un bras robotisé pour saisir des objets dans un entrepôt logistique.
* **Contexte 2 (Logistique / Gestion) :** Optimisation séquentielle de la gestion des niveaux de stocks de conteneurs dans un terminal portuaire.
* **Structure de données :** Grille d'états-actions (Q-Table).
* **Algorithme concurrent :** SARSA.
* **Situation d'inadaptation :** Espaces d'états ou d'actions continus infinis (car la table $Q$ devient impossible à stocker ; requiert alors des méthodes à approximation de fonctions comme DQN).
* **Métrique de performance :** Récompense moyenne accumulée par épisode.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Étudiez les récompenses des épisodes, pas seulement Q moyen.*
Nous allons analyser l'évolution de la somme des récompenses cumulées lors des derniers épisodes pour observer si l'agent trouve un chemin stable d'évitement du précipice.

````python
import gymnasium as gym

env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
historique_recompenses = []

for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    somme_rec = 0
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        somme_rec += recompense
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
    historique_recompenses.append(somme_rec)

print('Table Q Q-Learning :', q_table.shape, '| Valeur moyenne :', round(q_table.mean(), 2))
print(f"Moyenne des récompenses des 100 derniers épisodes : {np.mean(historique_recompenses[-100:]):.2f}")
env.close()
````

## 28. SARSA

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement (On-policy).
* **Entrées :** État courant $s_t$ et action choisie $a_t$.
* **Sortie :** Valeur d'action $Q(s, a)$.
* **Hypothèse clé :** L'agent met à jour ses valeurs d'action à partir de l'action réelle $a_{t+1}$ qui sera effectivement exécutée à l'étape suivante, ce qui le rend sensible aux risques liés aux actions d'exploration.
* **Étapes de calcul :** (1) Choix de $a$ depuis $s$ (politique $\epsilon$-greedy). (2) Transition vers $s'$ et choix de $a'$ (politique $\epsilon$-greedy). (3) Mise à jour On-policy. (4) $s \leftarrow s'$, $a \leftarrow a'$.
* **Paramètre appris :** Q-table $Q(s, a)$.
* **Hyperparamètre :** Taux d'exploration $\epsilon$ et taux d'apprentissage $\alpha$.

**Équation de mise à jour temporelle SARSA :**
$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R + \gamma Q(s', a') - Q(s, a) \right]$$
* **$Q(s', a')$ :** Valeur de l'action réelle choisie par la politique courante (On-Policy).

### C2 — Schéma original
```
    [ État s, Action a ] ===> Transition vers s' ===> Sélection de l'action réelle a'
                                                            |
         [ Mise à jour On-Policy en utilisant Q(s', a') ] <---
```

### C3 — Contextes et choix
* **Contexte 1 (Robotique Physique) :** Apprentissage de la marche d'un robot quadrupède où une chute peut casser le matériel (comportement prudent nécessaire pendant l'entraînement).
* **Contexte 2 (Gestion d'installations complexes) :** Conduite d'un réacteur chimique nécessitant d'éviter les surchauffes critiques dues à l'exploration.
* **Structure de données :** Grille d'états-actions (Q-Table).
* **Algorithme concurrent :** Q-learning.
* **Situation d'inadaptation :** Scénarios où la trajectoire exploratoire n'est pas modélisable ou change de façon chaotique.
* **Métrique de performance :** Récompense moyenne des épisodes de test.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?*
Oui. Dans l'environnement CliffWalking, le Q-learning cherche le chemin optimal géométrique direct (qui longe la falaise, risquant de tomber à la moindre exploration aléatoire), alors que SARSA apprend une trajectoire plus sécurisée et éloignée de la falaise. Nous allons évaluer la récompense moyenne finale pour le confirmer.

````python
import gymnasium as gym

env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
historique_recompenses_sarsa = []

def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))

for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    somme_rec = 0
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        somme_rec += recompense
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
    historique_recompenses_sarsa.append(somme_rec)

print('Table Q SARSA :', q_table.shape, '| Valeur moyenne :', round(q_table.mean(), 2))
print(f"Moyenne des récompenses SARSA (100 derniers épisodes) : {np.mean(historique_recompenses_sarsa[-100:]):.2f}")
env.close()
````

## 29. DQN — cible conceptuelle

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement basé sur les réseaux de neurones (Deep RL).
* **Entrées :** Caractéristiques d'état vectorielles de haute dimension.
* **Sortie :** Valeurs $Q(s, a)$ approchées pour toutes les actions possibles.
* **Hypothèse clé :** L'utilisation d'un réseau de neurones profond agissant comme approximateur de fonction non-linéaire universel permet de généraliser l'apprentissage des valeurs Q sur de grands espaces d'états.
* **Étapes de calcul :** (1) Sauvegarde des transitions dans une mémoire de rejeu (Experience Replay). (2) Tirage d'un mini-lot de transitions. (3) Calcul de la cible de Bellman à l'aide d'un réseau cible (Target Network). (4) Descente de gradient pour ajuster les poids du réseau d'évaluation.
* **Paramètres appris :** Les poids $\theta$ du réseau neuronal d'approximation.
* **Hyperparamètre :** Taille de la mémoire de rejeu, fréquence de mise à jour du réseau cible.

**Équation de perte de Bellman (MSE d'évaluation) :**
$$\mathcal{L}(\theta) = \mathbb{E} \left[ \left( R + \gamma \max_{a'} Q(s', a'; \theta^{-}) - Q(s, a; \theta) \right)^2 \right]$$
* **$\theta$ :** Poids du réseau de neurones d'évaluation.
* **$\theta^{-}$ :** Poids du réseau cible gelé.

### C2 — Schéma original
```
     [ État s ] ===> [ Réseau Neuronal Q(s, a; theta) ] ===> Valeurs Q estimées
                            ^
                            | (Ajustement par gradient de la perte de Bellman)
                            +--- [ Cible calculée : R + gamma * max_a' Q(s', a'; theta^-) ]
```

### C3 — Contextes et choix
* **Contexte 1 (Jeux Atari) :** Apprentissage autonome de jeux vidéo d'arcade à partir des pixels de l'écran (très haute dimension).
* **Contexte 2 (Automobile) :** Conduite autonome simulée d'un véhicule sur circuit urbain complexe.
* **Structure de données :** Tenseurs d'états d'entrée continus.
* **Algorithme concurrent :** Double DQN, Dueling DQN, ou PPO.
* **Situation d'inadaptation :** Environnements réels où l'entraînement direct est trop dangereux ou lent (requiert des milliards de pas d'entraînement).
* **Métrique de performance :** Score moyen accumulé par épisode.

### C4 — Test de sensibilité et interprétation
**Variante C — question spécifique :** *Pourquoi la cible est-elle égale à la récompense en état terminal ?*
En état terminal, il n'y a plus aucun état ultérieur possible ni aucune action future à entreprendre. Le terme d'actualisation de la valeur future $\gamma \max_{a'} Q(s', a')$ s'annule donc totalement, ce qui rend la cible de Bellman de cet échantillon strictement égale à la récompense immédiate reçue $R$.

````python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
````
